```python
import pandas as pd
import numpy as np
```

## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My task is mainly focused on ranking webpages by how likely they are to require a content refresh. A binary classification model can predict the probability that a page is experiencing a decline, and these probabilities can then be used to rank pages from highest to lowest priority for the SEO team to review.


## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target variable is **is_declining_label**, which is generated from the **trend_direction** column using a predefined rule. When a webpage is classified as declining, the label is assigned accordingly. The model then learns to predict this label based on the available page features rather than applying the rule directly.


## 3. Success metric

*One metric you can defend. What number means 'good'?*

The main evaluation metric is **Precision@50**. Since the SEO team can only review a limited number of pages, the model should prioritize the pages most likely to require attention. Precision@50 measures the percentage of the top 50 recommended pages that are actually declining. A higher score indicates that the model is effectively identifying the pages that need review the most.


## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is a single content page or webpage. Each row in the dataframe represents one webpage and includes SEO-related metrics such as clicks, impressions, CTR, ranking position, and other relevant features. The model generates one prediction for each page to determine its priority for review.

```python
# Create a synthetic DataFrame representing the unit of analysis (one row = one webpage)
data = {
    'page_id': range(1, 101),
    'clicks': np.random.randint(100, 5000, 100),
    'impressions': np.random.randint(1000, 50000, 100),
    'ctr': np.random.uniform(0.01, 0.15, 100).round(4),
    'ranking_position': np.random.randint(1, 50, 100),
    'search_volume': np.random.randint(500, 10000, 100),
    'user_engagement_score': np.random.uniform(0.1, 0.9, 100).round(2),
    'content_age_days': np.random.randint(30, 720, 100)
}
df = pd.DataFrame(data)

# Introduce 'trend_direction' and derive 'is_declining_label' as described in Section 2
def get_trend_direction():
    return np.random.choice(['stable', 'growing', 'declining'], p=[0.6, 0.2, 0.2])
df['trend_direction'] = [get_trend_direction() for _ in range(len(df))]
df['is_declining_label'] = (df['trend_direction'] == 'declining').astype(int)

# Display the first 5 rows to show the unit of analysis
print("DataFrame showing the unit of analysis (one row = one webpage):")
display(df.head())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Machine learning is a better choice than relying on a fixed rule because a webpage’s performance depends on many factors working together. Metrics like clicks, impressions, CTR, ranking position, search volume, user engagement, and content characteristics can all indicate whether a page needs to be refreshed. A simple rule, such as “if CTR is below 2%, refresh the page,” cannot capture these complex relationships and may lead to inaccurate recommendations. Machine learning, on the other hand, can learn from historical data and identify patterns across multiple features, helping the SEO team prioritize the pages that are most likely to need attention.
